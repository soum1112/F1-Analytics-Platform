```python
import fastf1
import pandas as pd

fastf1.Cache.enable_cache('../data')

session = fastf1.get_session(2025, 'Bahrain', 'R')
session.load()

print("Loaded!")
```

```python
# Grab results and keep only the useful columns
results = session.results

results_clean = results[[
    'DriverNumber',
    'Abbreviation',
    'FullName',
    'TeamName',
    'Position',
    'Points',
    'GridPosition',
    'Status'
]].copy()

# Add race info so you know which race this is
results_clean['Season'] = 2025
results_clean['Race'] = 'Bahrain'

print(results_clean)
```

```python
laps = session.laps
laps_clean = laps[[
    'Driver',
    'LapNumber',
    'LapTime',
    'Sector1Time',
    'Sector2Time',
    'Sector3Time',
    'Compound',
    'TyreLife',
    'IsPersonalBest'
]].copy()

# Convert lap times from timedelta to seconds (plain numbers are easier to work with)
laps_clean['LapTime_sec'] = laps_clean['LapTime'].dt.total_seconds()
laps_clean['Sector1_sec'] = laps_clean['Sector1Time'].dt.total_seconds()
laps_clean['Sector2_sec'] = laps_clean['Sector2Time'].dt.total_seconds()
laps_clean['Sector3_sec'] = laps_clean['Sector3Time'].dt.total_seconds()

# Add race info
laps_clean['Season'] = 2025
laps_clean['Race'] = 'Bahrain'

print(laps_clean.head(10))
```

```python
weather = session.weather_data
weather_clean = weather[[
    'Time',
    'AirTemp',
    'TrackTemp',
    'Humidity',
    'WindSpeed',
    'Rainfall'
]].copy()

weather_clean['Season'] = 2025
weather_clean['Race'] = 'Bahrain'

print(weather_clean.head(10))
```

```python
# Save to your data folder
results_clean.to_csv('../data/bahrain_2025_results.csv', index=False)
laps_clean.to_csv('../data/bahrain_2025_laps.csv', index=False)
weather_clean.to_csv('../data/bahrain_2025_weather.csv', index=False)

print("All files saved!")
```
